```python
# ============================================================
# 🌾 SEASONAL AGRICULTURE PERFORMANCE ANALYSIS - FULL PROJECT
# ============================================================

# -------------------- 1. IMPORT LIBRARIES --------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f_oneway
import os
import warnings

warnings.filterwarnings("ignore")

print("=" * 70)
print("🌾 SEASONAL AGRICULTURE PERFORMANCE ANALYSIS")
print("=" * 70)


# -------------------- 2. LOAD DATASET --------------------

FILE_PATH = "/content/spreadsheet.xlsx"

# Automatically detect CSV or Excel
if not os.path.exists(FILE_PATH):
    csv_files = [f for f in os.listdir() if f.endswith(".csv")]
    excel_files = [f for f in os.listdir() if f.endswith(".xlsx")]

    if csv_files:
        FILE_PATH = csv_files[0]
    elif excel_files:
        FILE_PATH = excel_files[0]
    else:
        raise FileNotFoundError(
            "❌ No CSV or Excel file found. Please upload your dataset first."
        )

if FILE_PATH.endswith(".csv"):
    df = pd.read_csv(FILE_PATH)
else:
    df = pd.read_excel(FILE_PATH)

print("\n✅ Dataset Loaded Successfully!")
print("📁 File:", FILE_PATH)


# -------------------- 3. DATASET OVERVIEW --------------------

print("\n" + "=" * 70)
print("1. DATASET OVERVIEW")
print("=" * 70)

print("\nDataset Shape:", df.shape)
print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
print(df.head())

print("\nDataset Information:")
df.info()


# -------------------- 4. STATISTICAL SUMMARY --------------------

print("\n" + "=" * 70)
print("2. STATISTICAL SUMMARY")
print("=" * 70)

print(df.describe(include="all"))


# -------------------- 5. MISSING VALUES --------------------

print("\n" + "=" * 70)
print("3. MISSING VALUE ANALYSIS")
print("=" * 70)

missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

print(missing_summary[missing_summary["Missing Values"] > 0])

# Fill all numerical missing values using median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

# Fill categorical missing values using mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing Values After Cleaning:")
print(df.isnull().sum().sum())


# -------------------- 6. DUPLICATE CHECK --------------------

print("\n" + "=" * 70)
print("4. DUPLICATE ANALYSIS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

if duplicates > 0:
    df = df.drop_duplicates()
    print("✅ Duplicate rows removed")

print("Final Dataset Shape:", df.shape)


# -------------------- 7. UNIQUE VALUES --------------------

print("\n" + "=" * 70)
print("5. CATEGORICAL VARIABLES")
print("=" * 70)

important_categorical_columns = [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]

for column in important_categorical_columns:
    if column in df.columns:
        print(f"\n{column}:")
        print("Unique Values:", df[column].nunique())
        print(df[column].unique())


# ============================================================
# 8. SEASON DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("6. SEASON DISTRIBUTION")
print("=" * 70)

season_count = df["Season"].value_counts()

print(season_count)

plt.figure(figsize=(8, 5))
plt.bar(season_count.index, season_count.values)
plt.title("Number of Farms in Each Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.show()


# ============================================================
# 9. CROP DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("7. CROP DISTRIBUTION")
print("=" * 70)

crop_count = df["Crop"].value_counts()

print(crop_count)

plt.figure(figsize=(10, 5))
plt.bar(crop_count.index, crop_count.values)
plt.title("Crop Distribution")
plt.xlabel("Crop")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# 10. STATE DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("8. STATE DISTRIBUTION")
print("=" * 70)

state_count = df["State"].value_counts()

print(state_count)

plt.figure(figsize=(10, 5))
plt.bar(state_count.index, state_count.values)
plt.title("Number of Farms by State")
plt.xlabel("State")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# 11. SEASON-WISE PERFORMANCE
# ============================================================

print("\n" + "=" * 70)
print("9. SEASON-WISE AGRICULTURAL PERFORMANCE")
print("=" * 70)

season_summary = df.groupby("Season").agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Production=("Production_Tonnes", "mean"),
    Average_Rainfall=("Rainfall_mm", "mean"),
    Average_Temperature=("Avg_Temperature_C", "mean"),
    Average_Humidity=("Humidity_pct", "mean"),
    Average_Soil_Moisture=("Soil_Moisture_pct", "mean"),
    Average_Fertilizer=("Fertilizer_kg_ha", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Disease_Risk=("Disease_Pest_Risk_pct", "mean")
).round(2)

print(season_summary)


# -------------------- Yield by Season --------------------

season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_yield.index, season_yield.values)
plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Hectare)")
plt.show()


# -------------------- Production by Season --------------------

season_production = df.groupby("Season")["Production_Tonnes"].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_production.index, season_production.values)
plt.title("Average Production by Season")
plt.xlabel("Season")
plt.ylabel("Production (Tonnes)")
plt.show()


# -------------------- Profit by Season --------------------

season_profit = df.groupby("Season")["Profit_INR"].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_profit.index, season_profit.values)
plt.axhline(0)
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.show()


# -------------------- Revenue by Season --------------------

season_revenue = df.groupby("Season")["Revenue_INR"].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_revenue.index, season_revenue.values)
plt.title("Average Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue (INR)")
plt.show()


# ============================================================
# 12. ENVIRONMENTAL ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("10. ENVIRONMENTAL CONDITIONS BY SEASON")
print("=" * 70)

environmental_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct"
]

environment_summary = df.groupby("Season")[
    environmental_columns
].mean().round(2)

print(environment_summary)


# Rainfall

plt.figure(figsize=(8, 5))
plt.bar(
    df.groupby("Season")["Rainfall_mm"].mean().index,
    df.groupby("Season")["Rainfall_mm"].mean().values
)
plt.title("Average Rainfall by Season")
plt.xlabel("Season")
plt.ylabel("Rainfall (mm)")
plt.show()


# Temperature

plt.figure(figsize=(8, 5))
plt.bar(
    df.groupby("Season")["Avg_Temperature_C"].mean().index,
    df.groupby("Season")["Avg_Temperature_C"].mean().values
)
plt.title("Average Temperature by Season")
plt.xlabel("Season")
plt.ylabel("Temperature (°C)")
plt.show()


# Humidity

plt.figure(figsize=(8, 5))
plt.bar(
    df.groupby("Season")["Humidity_pct"].mean().index,
    df.groupby("Season")["Humidity_pct"].mean().values
)
plt.title("Average Humidity by Season")
plt.xlabel("Season")
plt.ylabel("Humidity (%)")
plt.show()


# Soil Moisture

plt.figure(figsize=(8, 5))
plt.bar(
    df.groupby("Season")["Soil_Moisture_pct"].mean().index,
    df.groupby("Season")["Soil_Moisture_pct"].mean().values
)
plt.title("Average Soil Moisture by Season")
plt.xlabel("Season")
plt.ylabel("Soil Moisture (%)")
plt.show()


# ============================================================
# 13. RESOURCE USAGE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("11. RESOURCE USAGE BY SEASON")
print("=" * 70)

resource_columns = [
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Water_Used_m3"
]

resource_summary = df.groupby("Season")[
    resource_columns
].mean().round(2)

print(resource_summary)


for column in [
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha"
]:

    season_resource = df.groupby("Season")[column].mean()

    plt.figure(figsize=(8, 5))

    plt.bar(
        season_resource.index,
        season_resource.values
    )

    plt.title(f"{column} by Season")
    plt.xlabel("Season")
    plt.ylabel(column)

    plt.show()


# ============================================================
# 14. IRRIGATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("12. IRRIGATION ANALYSIS")
print("=" * 70)

irrigation_count = df["Irrigation_Method"].value_counts()

print("\nIrrigation Distribution:")
print(irrigation_count)

plt.figure(figsize=(8, 5))
plt.bar(irrigation_count.index, irrigation_count.values)
plt.title("Irrigation Method Distribution")
plt.xlabel("Irrigation Method")
plt.ylabel("Number of Farms")
plt.show()


irrigation_summary = df.groupby(
    "Irrigation_Method"
).agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Water_Used=("Water_Used_m3", "mean"),
    Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
).round(2)

print("\nIrrigation Performance:")
print(irrigation_summary)


# Yield by Irrigation

plt.figure(figsize=(8, 5))

plt.bar(
    irrigation_summary.index,
    irrigation_summary["Average_Yield"]
)

plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Yield (Tonnes/Hectare)")

plt.show()


# Water Efficiency

plt.figure(figsize=(8, 5))

plt.bar(
    irrigation_summary.index,
    irrigation_summary["Water_Efficiency"]
)

plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency")

plt.show()


# ============================================================
# 15. CROP-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("13. CROP-WISE ANALYSIS")
print("=" * 70)

crop_summary = df.groupby(
    "Crop"
).agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Production=("Production_Tonnes", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Disease_Risk=("Disease_Pest_Risk_pct", "mean")
).round(2)

print(crop_summary)


# Yield by Crop

crop_yield = df.groupby(
    "Crop"
)["Yield_Tonnes_Ha"].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    crop_yield.index,
    crop_yield.values
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Yield (Tonnes/Hectare)")

plt.xticks(rotation=45)

plt.show()


# Profit by Crop

crop_profit = df.groupby(
    "Crop"
)["Profit_INR"].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    crop_profit.index,
    crop_profit.values
)

plt.axhline(0)

plt.title("Average Profit by Crop")
plt.xlabel("Crop")
plt.ylabel("Profit (INR)")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 16. CROP + SEASON ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("14. CROP AND SEASON ANALYSIS")
print("=" * 70)

crop_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
).round(2)

print("\nAverage Yield by Crop and Season:")
print(crop_season_yield)


plt.figure(figsize=(10, 6))

plt.imshow(crop_season_yield)

plt.colorbar()

plt.xticks(
    range(len(crop_season_yield.columns)),
    crop_season_yield.columns
)

plt.yticks(
    range(len(crop_season_yield.index)),
    crop_season_yield.index
)

plt.title("Yield by Crop and Season")

plt.xlabel("Season")
plt.ylabel("Crop")

plt.show()


# Crop + Season Profit

crop_season_profit = pd.pivot_table(
    df,
    values="Profit_INR",
    index="Crop",
    columns="Season",
    aggfunc="mean"
).round(2)

print("\nAverage Profit by Crop and Season:")
print(crop_season_profit)


plt.figure(figsize=(10, 6))

plt.imshow(crop_season_profit)

plt.colorbar()

plt.xticks(
    range(len(crop_season_profit.columns)),
    crop_season_profit.columns
)

plt.yticks(
    range(len(crop_season_profit.index)),
    crop_season_profit.index
)

plt.title("Profit by Crop and Season")

plt.xlabel("Season")
plt.ylabel("Crop")

plt.show()


# ============================================================
# 17. STATE-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("15. STATE-WISE ANALYSIS")
print("=" * 70)

state_summary = df.groupby(
    "State"
).agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Revenue=("Revenue_INR", "mean")
).round(2)

print(state_summary)


state_yield = df.groupby(
    "State"
)["Yield_Tonnes_Ha"].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    state_yield.index,
    state_yield.values
)

plt.title("Average Yield by State")

plt.xlabel("State")
plt.ylabel("Yield (Tonnes/Hectare)")

plt.xticks(rotation=45)

plt.show()


# State + Season

state_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="State",
    columns="Season",
    aggfunc="mean"
).round(2)

print("\nState-wise Seasonal Yield:")
print(state_season_yield)


plt.figure(figsize=(10, 6))

plt.imshow(state_season_yield)

plt.colorbar()

plt.xticks(
    range(len(state_season_yield.columns)),
    state_season_yield.columns
)

plt.yticks(
    range(len(state_season_yield.index)),
    state_season_yield.index
)

plt.title("State-wise Seasonal Yield")

plt.xlabel("Season")
plt.ylabel("State")

plt.show()


# ============================================================
# 18. ECONOMIC ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("16. ECONOMIC ANALYSIS")
print("=" * 70)


# Create Profitability Status

df["Performance_Status"] = np.where(
    df["Profit_INR"] > 0,
    "Profitable",
    "Loss"
)


# Create Profit Margin

df["Profit_Margin_pct"] = np.where(
    df["Revenue_INR"] != 0,
    (df["Profit_INR"] / df["Revenue_INR"]) * 100,
    0
)


# Revenue per Hectare

df["Revenue_Per_Hectare"] = (
    df["Revenue_INR"] /
    df["Farm_Area_Hectares"]
)


# Profit per Hectare

df["Profit_Per_Hectare"] = (
    df["Profit_INR"] /
    df["Farm_Area_Hectares"]
)


performance_count = df[
    "Performance_Status"
].value_counts()

print("\nProfitable vs Loss Farms:")
print(performance_count)


plt.figure(figsize=(7, 5))

plt.bar(
    performance_count.index,
    performance_count.values
)

plt.title("Profitable vs Loss-Making Farms")

plt.xlabel("Farm Status")
plt.ylabel("Number of Farms")

plt.show()


# Profit Distribution

plt.figure(figsize=(10, 6))

plt.hist(
    df["Profit_INR"],
    bins=30
)

plt.axvline(0)

plt.title("Profit Distribution")

plt.xlabel("Profit (INR)")
plt.ylabel("Number of Farms")

plt.show()


# Revenue vs Cost

plt.figure(figsize=(8, 6))

plt.scatter(
    df["Total_Cost_INR"],
    df["Revenue_INR"],
    alpha=0.5
)

plt.title("Total Cost vs Revenue")

plt.xlabel("Total Cost (INR)")
plt.ylabel("Revenue (INR)")

plt.show()


# Profitability by Season

season_profitability = pd.crosstab(
    df["Season"],
    df["Performance_Status"]
)

print("\nProfitability by Season:")
print(season_profitability)


season_profitability.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Profitability by Season")

plt.xlabel("Season")
plt.ylabel("Number of Farms")

plt.xticks(rotation=0)

plt.show()


# ============================================================
# 19. WATER EFFICIENCY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("17. WATER EFFICIENCY ANALYSIS")
print("=" * 70)

water_season = df.groupby(
    "Season"
)["Water_Efficiency_t_per_1000m3"].mean()

print("\nWater Efficiency by Season:")
print(water_season)


plt.figure(figsize=(8, 5))

plt.bar(
    water_season.index,
    water_season.values
)

plt.title("Water Efficiency by Season")

plt.xlabel("Season")
plt.ylabel("Water Efficiency")

plt.show()


crop_water = df.groupby(
    "Crop"
)["Water_Efficiency_t_per_1000m3"].mean().sort_values(
    ascending=False
)

print("\nWater Efficiency by Crop:")
print(crop_water)


plt.figure(figsize=(10, 5))

plt.bar(
    crop_water.index,
    crop_water.values
)

plt.title("Water Efficiency by Crop")

plt.xlabel("Crop")
plt.ylabel("Water Efficiency")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 20. DISEASE AND PEST ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("18. DISEASE AND PEST ANALYSIS")
print("=" * 70)

disease_season = df.groupby(
    "Season"
)["Disease_Pest_Risk_pct"].mean()

print(disease_season)


plt.figure(figsize=(8, 5))

plt.bar(
    disease_season.index,
    disease_season.values
)

plt.title("Average Disease/Pest Risk by Season")

plt.xlabel("Season")
plt.ylabel("Risk (%)")

plt.show()


# Disease Risk vs Yield

plt.figure(figsize=(8, 6))

plt.scatter(
    df["Disease_Pest_Risk_pct"],
    df["Yield_Tonnes_Ha"],
    alpha=0.5
)

plt.title("Disease/Pest Risk vs Yield")

plt.xlabel("Disease/Pest Risk (%)")
plt.ylabel("Yield")

plt.show()


# Disease Risk vs Profit

plt.figure(figsize=(8, 6))

plt.scatter(
    df["Disease_Pest_Risk_pct"],
    df["Profit_INR"],
    alpha=0.5
)

plt.axhline(0)

plt.title("Disease/Pest Risk vs Profit")

plt.xlabel("Disease/Pest Risk (%)")
plt.ylabel("Profit (INR)")

plt.show()


# ============================================================
# 21. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("19. CORRELATION ANALYSIS")
print("=" * 70)

numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()


# Correlation with Yield

yield_correlation = correlation_matrix[
    "Yield_Tonnes_Ha"
].sort_values(
    ascending=False
)

print("\nCorrelation with Yield:")
print(yield_correlation)


yield_correlation_plot = yield_correlation.drop(
    "Yield_Tonnes_Ha"
)

plt.figure(figsize=(10, 10))

plt.barh(
    yield_correlation_plot.index,
    yield_correlation_plot.values
)

plt.title("Correlation of Variables with Yield")

plt.xlabel("Correlation")

plt.show()


# Correlation with Profit

profit_correlation = correlation_matrix[
    "Profit_INR"
].sort_values(
    ascending=False
)

print("\nCorrelation with Profit:")
print(profit_correlation)


profit_correlation_plot = profit_correlation.drop(
    "Profit_INR"
)

plt.figure(figsize=(10, 10))

plt.barh(
    profit_correlation_plot.index,
    profit_correlation_plot.values
)

plt.title("Correlation of Variables with Profit")

plt.xlabel("Correlation")

plt.show()


# ============================================================
# 22. ENVIRONMENT VS YIELD
# ============================================================

print("\n" + "=" * 70)
print("20. ENVIRONMENTAL FACTORS VS YIELD")
print("=" * 70)

environment_vs_yield = [
    ("Rainfall_mm", "Rainfall (mm)"),
    ("Avg_Temperature_C", "Temperature (°C)"),
    ("Humidity_pct", "Humidity (%)"),
    ("Sunlight_Hours_Day", "Sunlight Hours"),
    ("Soil_Moisture_pct", "Soil Moisture (%)"),
    ("Fertilizer_kg_ha", "Fertilizer (kg/ha)")
]

for column, label in environment_vs_yield:

    plt.figure(figsize=(8, 6))

    plt.scatter(
        df[column],
        df["Yield_Tonnes_Ha"],
        alpha=0.5
    )

    plt.title(f"{label} vs Yield")

    plt.xlabel(label)

    plt.ylabel("Yield (Tonnes/Hectare)")

    plt.show()


# ============================================================
# 23. TOP PERFORMING FARMS
# ============================================================

print("\n" + "=" * 70)
print("21. TOP PERFORMING FARMS")
print("=" * 70)


print("\n🏆 TOP 10 FARMS BY YIELD")

top_yield_farms = df.nlargest(
    10,
    "Yield_Tonnes_Ha"
)

print(
    top_yield_farms[
        [
            "Farm_ID",
            "State",
            "Crop",
            "Season",
            "Yield_Tonnes_Ha",
            "Profit_INR"
        ]
    ]
)


print("\n💰 TOP 10 MOST PROFITABLE FARMS")

top_profit_farms = df.nlargest(
    10,
    "Profit_INR"
)

print(
    top_profit_farms[
        [
            "Farm_ID",
            "State",
            "Crop",
            "Season",
            "Yield_Tonnes_Ha",
            "Profit_INR"
        ]
    ]
)


print("\n📉 TOP 10 LOSS-MAKING FARMS")

top_loss_farms = df.nsmallest(
    10,
    "Profit_INR"
)

print(
    top_loss_farms[
        [
            "Farm_ID",
            "State",
            "Crop",
            "Season",
            "Yield_Tonnes_Ha",
            "Profit_INR"
        ]
    ]
)


# ============================================================
# 24. OUTLIER DETECTION
# ============================================================

print("\n" + "=" * 70)
print("22. OUTLIER DETECTION")
print("=" * 70)


def detect_outliers(data, column):

    Q1 = data[column].quantile(0.25)

    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return outliers


outlier_columns = [
    "Yield_Tonnes_Ha",
    "Profit_INR",
    "Rainfall_mm",
    "Water_Used_m3",
    "Production_Tonnes"
]


for column in outlier_columns:

    outliers = detect_outliers(
        df,
        column
    )

    print(
        f"\n{column} Outliers:",
        len(outliers)
    )


# ============================================================
# 25. STATISTICAL TESTING - ANOVA
# ============================================================

print("\n" + "=" * 70)
print("23. STATISTICAL ANALYSIS - ANOVA")
print("=" * 70)


# Yield ANOVA

season_groups_yield = [
    group["Yield_Tonnes_Ha"].dropna().values
    for name, group in df.groupby("Season")
]

f_stat_yield, p_value_yield = f_oneway(
    *season_groups_yield
)

print("\nANOVA TEST: YIELD BY SEASON")

print("F Statistic:", round(f_stat_yield, 4))

print("P Value:", p_value_yield)

if p_value_yield < 0.05:

    print(
        "✅ Result: Yield differs significantly across seasons."
    )

else:

    print(
        "❌ Result: No statistically significant difference in yield."
    )


# Profit ANOVA

season_groups_profit = [
    group["Profit_INR"].dropna().values
    for name, group in df.groupby("Season")
]

f_stat_profit, p_value_profit = f_oneway(
    *season_groups_profit
)

print("\nANOVA TEST: PROFIT BY SEASON")

print("F Statistic:", round(f_stat_profit, 4))

print("P Value:", p_value_profit)

if p_value_profit < 0.05:

    print(
        "✅ Result: Profit differs significantly across seasons."
    )

else:

    print(
        "❌ Result: No statistically significant difference in profit."
    )


# ============================================================
# 26. BEST PERFORMERS
# ============================================================

print("\n" + "=" * 70)
print("24. BEST AGRICULTURAL PERFORMERS")
print("=" * 70)


best_season_yield = df.groupby(
    "Season"
)["Yield_Tonnes_Ha"].mean().idxmax()


best_season_profit = df.groupby(
    "Season"
)["Profit_INR"].mean().idxmax()


best_crop_yield = df.groupby(
    "Crop"
)["Yield_Tonnes_Ha"].mean().idxmax()


best_crop_profit = df.groupby(
    "Crop"
)["Profit_INR"].mean().idxmax()


best_irrigation_yield = df.groupby(
    "Irrigation_Method"
)["Yield_Tonnes_Ha"].mean().idxmax()


best_irrigation_water = df.groupby(
    "Irrigation_Method"
)["Water_Efficiency_t_per_1000m3"].mean().idxmax()


print("\n🌦️ Best Season by Yield:", best_season_yield)

print("💰 Best Season by Profit:", best_season_profit)

print("🌾 Best Crop by Yield:", best_crop_yield)

print("💵 Most Profitable Crop:", best_crop_profit)

print(
    "💧 Best Irrigation Method by Yield:",
    best_irrigation_yield
)

print(
    "💧 Best Irrigation Method by Water Efficiency:",
    best_irrigation_water
)


# ============================================================
# 27. BEST CROP IN EACH SEASON
# ============================================================

print("\n" + "=" * 70)
print("25. BEST CROP IN EACH SEASON")
print("=" * 70)


crop_season_performance = df.groupby(
    ["Season", "Crop"]
).agg(
    Average_Yield=(
        "Yield_Tonnes_Ha",
        "mean"
    ),

    Average_Profit=(
        "Profit_INR",
        "mean"
    )
).reset_index()


best_yield_each_season = crop_season_performance.loc[
    crop_season_performance.groupby(
        "Season"
    )["Average_Yield"].idxmax()
]


best_profit_each_season = crop_season_performance.loc[
    crop_season_performance.groupby(
        "Season"
    )["Average_Profit"].idxmax()
]


print("\n🏆 Best Crop by Yield in Each Season:")

print(best_yield_each_season)


print("\n💰 Most Profitable Crop in Each Season:")

print(best_profit_each_season)


# ============================================================
# 28. FINAL PROJECT DASHBOARD
# ============================================================

print("\n" + "=" * 70)
print("🌾 FINAL PROJECT DASHBOARD")
print("=" * 70)

print("\n📊 DATASET SUMMARY")

print("Total Farms:", len(df))

print("Total States:", df["State"].nunique())

print("Total Districts:", df["District"].nunique())

print("Total Crops:", df["Crop"].nunique())

print("Total Seasons:", df["Season"].nunique())


print("\n🌾 PERFORMANCE SUMMARY")

print(
    "Best Season by Yield:",
    best_season_yield
)

print(
    "Best Season by Profit:",
    best_season_profit
)

print(
    "Best Crop by Yield:",
    best_crop_yield
)

print(
    "Most Profitable Crop:",
    best_crop_profit
)


print("\n💧 RESOURCE SUMMARY")

print(
    "Best Irrigation Method by Yield:",
    best_irrigation_yield
)

print(
    "Best Irrigation Method by Water Efficiency:",
    best_irrigation_water
)


print("\n📈 STATISTICAL RESULTS")

print(
    "Yield ANOVA P-Value:",
    p_value_yield
)

print(
    "Profit ANOVA P-Value:",
    p_value_profit
)


# ============================================================
# 29. KEY FINDINGS - AUTOMATICALLY GENERATED
# ============================================================

print("\n" + "=" * 70)
print("🔍 KEY FINDINGS")
print("=" * 70)


best_yield_value = df.groupby(
    "Season"
)["Yield_Tonnes_Ha"].mean().max()


best_profit_value = df.groupby(
    "Season"
)["Profit_INR"].mean().max()


print(
    f"""
1. The dataset contains {len(df)} agricultural farm records.

2. The season with the highest average yield is
   {best_season_yield} with an average yield of
   {best_yield_value:.2f} tonnes/hectare.

3. The season with the highest average profit is
   {best_season_profit} with an average profit of
   ₹{best_profit_value:,.2f}.

4. The crop with the highest average yield is
   {best_crop_yield}.

5. The most profitable crop is
   {best_crop_profit}.

6. The irrigation method producing the highest
   average yield is {best_irrigation_yield}.

7. The most water-efficient irrigation method is
   {best_irrigation_water}.

8. Statistical testing indicates whether seasonal
   differences in yield and profit are significant.
"""
)


# ============================================================
# 30. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "cleaned_seasonal_agriculture_data.csv",
    index=False
)


print("\n" + "=" * 70)
print("💾 FILE SAVED SUCCESSFULLY")
print("=" * 70)

print(
    "✅ Cleaned Dataset Saved As:"
)

print(
    "cleaned_seasonal_agriculture_data.csv"
)


# ============================================================
# 31. EXPORT IMPORTANT RESULTS TO EXCEL
# ============================================================

try:

    with pd.ExcelWriter(
        "Agriculture_Analysis_Results.xlsx"
    ) as writer:

        season_summary.to_excel(
            writer,
            sheet_name="Season Summary"
        )

        crop_summary.to_excel(
            writer,
            sheet_name="Crop Summary"
        )

        irrigation_summary.to_excel(
            writer,
            sheet_name="Irrigation Summary"
        )

        state_summary.to_excel(
            writer,
            sheet_name="State Summary"
        )

        crop_season_yield.to_excel(
            writer,
            sheet_name="Crop Season Yield"
        )

        crop_season_profit.to_excel(
            writer,
            sheet_name="Crop Season Profit"
        )

        yield_correlation.to_excel(
            writer,
            sheet_name="Yield Correlation"
        )

        profit_correlation.to_excel(
            writer,
            sheet_name="Profit Correlation"
        )

    print(
        "\n✅ Excel Results Saved As:"
    )

    print(
        "Agriculture_Analysis_Results.xlsx"
    )

except Exception as e:

    print(
        "\nExcel export skipped:",
        e
    )


print("\n" + "=" * 70)
print("=" * 70)
```
